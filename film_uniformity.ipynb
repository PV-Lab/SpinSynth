```python
import numpy as np
import pandas as pd
```

```python
data_path = '/Users/shengfang/Desktop/TRI/test_FAPbI3/HSI_test_FAPbI3/third_round_BO_20251002/'
sum_variance_list = []
for idx in range(0,6):
    df = pd.read_csv(f'{data_path}/pixel_reflectance_{idx}.csv',sep=',',header=0,index_col=0)
    sub_df = df.iloc[:, 1:]
    normalized_df = sub_df.div(sub_df.max(axis=0), axis=1)
    variance = normalized_df.var(axis=1)
    sum_variance = variance.sum()
    sum_variance_list.append(sum_variance)

print(sum_variance_list)
sum_variance_df = pd.DataFrame(sum_variance_list, columns=['Sum Variance'])
sum_variance_df.to_csv(f'{data_path}/sum_variance.csv', index=False)
```
